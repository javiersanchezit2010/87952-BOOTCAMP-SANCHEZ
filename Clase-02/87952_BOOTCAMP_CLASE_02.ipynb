```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np

# --- Datos de ejemplo ---
categorias = ["Python", "SQL", "Excel", "Power BI", "Machine Learning"]
valores = [92, 78, 65, 71, 58]
colores = ["#2E86AB", "#A23B72", "#F18F01", "#C73E1D", "#3B1F2B"]

# --- Estilo general ---
plt.style.use("seaborn-v0_8-whitegrid")
fig, ax = plt.subplots(figsize=(10, 6), dpi=120)
fig.patch.set_facecolor("#f7f7f7")
ax.set_facecolor("#f7f7f7")

# --- Barras horizontales con degradado de color ---
bars = ax.barh(categorias, valores, color=colores, edgecolor="black", linewidth=0.6, height=0.6)

# --- Etiquetas de valor al final de cada barra ---
for bar, valor in zip(bars, valores):
    ax.text(bar.get_width() + 1.5, bar.get_y() + bar.get_height()/2,
             f"{valor}%", va="center", fontsize=12, fontweight="bold", color="#333333")

# --- Títulos y ejes ---
ax.set_title("Dominio de Habilidades Técnicas del Curso",
              fontsize=18, fontweight="bold", color="#222222", pad=20)
ax.set_xlabel("Nivel de dominio (%)", fontsize=12, labelpad=10)
ax.set_xlim(0, 100)
ax.xaxis.set_major_formatter(mticker.PercentFormatter())

# --- Quitar bordes innecesarios ---
for spine in ["top", "right", "left"]:
    ax.spines[spine].set_visible(False)

ax.invert_yaxis()  # la primera categoría arriba
ax.tick_params(axis="y", labelsize=13)
ax.tick_params(axis="x", labelsize=11)

# --- Nota al pie ---
fig.text(0.13, 0.02, "Fuente: Datos simulados con fines educativos • matplotlib",
          fontsize=9, color="gray", style="italic")

plt.tight_layout()
plt.show()
```
